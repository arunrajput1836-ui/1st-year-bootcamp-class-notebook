```python
from bokeh.plotting import figure, show, output_file
import warnings
warnings.filterwarnings('ignore')
```

```python
from bokeh.plotting import  figure, show, output_file

output_file("line_chart.html")

p = figure(title="Simple Line Chart",
           x_axis_label = 'x Values',
           y_axis_label='y Values')

p.line([1,2,3,4,5],[6,7,2,5,7],
line_width=2,
color='green')

show(p)
```

```python
 from bokeh.plotting import  figure, show, output_file

output_file("Scatter_plot.html")

p = figure(title="Scatter plot Example",
           x_axis_label = 'x Values',
           y_axis_label='y Values')

p.circle([1,2,3,4,5],[6,7,2,5,7],
         size = 10,
line_width=2,
color='red',
         legend_label="Data Points")

show(p)
```

```python
from bokeh.plotting import  figure, show, output_file

output_file("bar_chart.html")


fruits =['apples','bananas','grapes','oranges']
counts =[10,15,7,12]
p = figure(x_range=fruits, title="fruits counts")

p.vbar(x=fruits, top=counts, width=0.5, color="skyblue")

p.xgrid.grid_line_color = None
p.ygrid.grid_line_color = None
p.y_range.start = 0



show(p)
```

```python
from bokeh.models import HoverTool
output_file("Hover_Tool.html")
p = figure(title="Hover Example", x_axis_label='x', y_axis_label='y')

x=[1,2,3,4,5]
y=[6,7,2,4,5]
p.circle(x,y, size=10, color='navy')

hover = HoverTool(tooltips=[("x value","@x"),("y value","@y")])
p.add_tools(hover)

show(p)
```
